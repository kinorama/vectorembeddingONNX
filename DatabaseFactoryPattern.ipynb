Factory pattern for database python

1. Define a common interface or abstract base class for database connections:

```python
from abc import ABC, abstractmethod

class DatabaseConnection(ABC):
    @abstractmethod
    def connect(self):
        pass

    @abstractmethod
    def execute_query(self, query):
        pass

    @abstractmethod
    def close(self):
        pass
```

2. Create concrete database connection classes:
   Each class implements the DatabaseConnection interface for a specific database type.

```python
class MySQLConnection(DatabaseConnection):
    def __init__(self, host, user, password, database):
        self.host = host
        self.user = user
        self.password = password
        self.database = database
        self._connection = None

    def connect(self):
        print(f"Connecting to MySQL database: {self.database} at {self.host}")
        # Simulate actual connection
        self._connection = "MySQL Connection Object" 

    def execute_query(self, query):
        if self._connection:
            print(f"Executing MySQL query: {query}")
            # Simulate query execution
            return f"MySQL Result for: {query}"
        else:
            print("Not connected to MySQL.")
            return None

    def close(self):
        if self._connection:
            print("Closing MySQL connection.")
            self._connection = None

class PostgreSQLConnection(DatabaseConnection):
    def __init__(self, host, user, password, database):
        self.host = host
        self.user = user
        self.password = password
        self.database = database
        self._connection = None

    def connect(self):
        print(f"Connecting to PostgreSQL database: {self.database} at {self.host}")
        # Simulate actual connection
        self._connection = "PostgreSQL Connection Object"

    def execute_query(self, query):
        if self._connection:
            print(f"Executing PostgreSQL query: {query}")
            # Simulate query execution
            return f"PostgreSQL Result for: {query}"
        else:
            print("Not connected to PostgreSQL.")
            return None

    def close(self):
        if self._connection:
            print("Closing PostgreSQL connection.")
            self._connection = None
```

3. Implement the Factory class:
   This class contains a method to create and return the appropriate DatabaseConnection object based on input parameters.

```python
class DatabaseConnectionFactory:
    @staticmethod
    def get_connection(db_type, **kwargs):
        if db_type == "mysql":
            return MySQLConnection(**kwargs)
        elif db_type == "postgresql":
            return PostgreSQLConnection(**kwargs)
        else:
            raise ValueError(f"Unsupported database type: {db_type}")
```

4. Client Code:
   The client code interacts with the factory to obtain a connection, without needing to know the specific concrete class.

```python
if __name__ == "__main__":
    # Get a MySQL connection
    mysql_conn = DatabaseConnectionFactory.get_connection(
        "mysql", host="localhost", user="root", password="password", database="mydb"
    )
    mysql_conn.connect()
    result_mysql = mysql_conn.execute_query("SELECT * FROM users;")
    print(result_mysql)
    mysql_conn.close()

    print("-" * 20)

    # Get a PostgreSQL connection
    pg_conn = DatabaseConnectionFactory.get_connection(
        "postgresql", host="localhost", user="admin", password="pg_password", database="otherdb"
    )
    pg_conn.connect()
    result_pg = pg_conn.execute_query("INSERT INTO products VALUES (1, 'Laptop');")
    print(result_pg)
    pg_conn.close()
```
